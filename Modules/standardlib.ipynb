```python
#list of even numbers between start to end
import sys

def even_numbers(start = 1, end = 20):
    result = []
    for index in range(start, end+1):
        if index % 2 == 0:
            result.append(index)
    return result
```

```python
even_numbers()
```

```python
#generator

def even_numbers_generator(start=1, end=10):
    for index in range(start, end+1):
        if index % 2 == 0:
            yield index
```

```python
even_numbers_generator()
```

```python
list(even_numbers_generator())
```

```python
for even_numbers in even_numbers_generator():
    print(even_numbers)
```
